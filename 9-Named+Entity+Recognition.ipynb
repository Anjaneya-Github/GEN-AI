```python
import nltk
from nltk.tokenize import word_tokenize

# 1. Ensure all necessary resources are downloaded
def download_ner_resources():
    resources = [
        'punkt', 
        'punkt_tab',
        'averaged_perceptron_tagger', 
        'averaged_perceptron_tagger_eng',
        'maxent_ne_chunker', 
        'maxent_ne_chunker_tab',
        'words'
    ]
    for resource in resources:
        try:
            nltk.data.find(f'help/taggers/{resource}') # Path varies, simple try/except is safer
        except LookupError:
            nltk.download(resource, quiet=True)
        except ValueError:
            nltk.download(resource, quiet=True)

download_ner_resources()

# 2. Prepare Data
sentence = "Elon Musk founded SpaceX in California."

# 3. Tokenize and POS Tag (Prerequisites for ne_chunk)
tokens = word_tokenize(sentence)
tag_elements = nltk.pos_tag(tokens)

# 4. Perform Named Entity Chunking
# binary=True groups all entities as 'NE'. 
# binary=False (default) attempts to categorize them (PERSON, ORGANIZATION, GPE, etc.)
ne_tree = nltk.ne_chunk(tag_elements, binary=False)

print(ne_tree)

# 5. Draw the tree
# Note: .draw() opens a separate window. In Jupyter, it might freeze if no GUI backend is available.
try:
    ne_tree.draw()
except Exception as e:
    print(f"Could not draw tree visually: {e}")
```

```python
#Named Entity Recognition (NER) is the process of locating and classifying named entities in text into predefined categories such as names of persons, organizations, locations, expressions of times, quantities, monetary values, percentages, etc.

#In NLTK, this is a three-step pipeline:

#Tokenization: Split sentences into words.
#POS Tagging: Assign parts of speech to those words.
#Chunking: Use the POS tags to identify entity chunks.
```

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
from nltk.tree import Tree

# ---------------------------------------------------------
# 1. Setup: Ensure all models are downloaded
# ---------------------------------------------------------
def download_ner_resources():
    resources = [
        'punkt', 
        'punkt_tab',                     # Tokenizer models
        'averaged_perceptron_tagger',    # POS Tagger
        'averaged_perceptron_tagger_eng',# POS Tagger (newer NLTK)
        'maxent_ne_chunker',             # NER Chunker
        'maxent_ne_chunker_tab',         # NER Chunker (newer NLTK)
        'words'                          # Vocabulary corpus
    ]
    
    print("Verifying NLTK resources...")
    for res in resources:
        try:
            nltk.data.find(f'help/taggers/{res}')
        except (LookupError, ValueError):
            # Quietly download if missing
            nltk.download(res, quiet=True)
    print("Resources ready.")

download_ner_resources()

# ---------------------------------------------------------
# 2. The NER Function
# ---------------------------------------------------------
def extract_named_entities(text):
    # Step 1: Tokenize
    tokens = word_tokenize(text)
    
    # Step 2: POS Tagging
    tags = pos_tag(tokens)
    
    # Step 3: NER Chunking
    # binary=False returns specific categories (Person, Organization, GPE)
    # binary=True returns just "NE" (Named Entity) for everything
    tree = ne_chunk(tags, binary=False)
    
    entities = []
    
    # Step 4: Extracting Data from the Tree
    # The tree contains both tuples (word, pos) and Tree objects (Entity, [words])
    for subtree in tree:
        if isinstance(subtree, Tree):
            entity_name = " ".join([word for word, tag in subtree.leaves()])
            entity_type = subtree.label()
            entities.append((entity_name, entity_type))
            
    return entities

# ---------------------------------------------------------
# 3. Execution
# ---------------------------------------------------------
text = "Sundar Pichai works for Google in California. He announced new AI features on Monday."

results = extract_named_entities(text)

print(f"Input Text: {text}\n")
print("Found Entities:")
for name, category in results:
    print(f" - {name} ({category})")
```
