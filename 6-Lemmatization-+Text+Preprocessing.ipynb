## Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −


```python
## Q&A,chatbots,text summarization
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Ensure resources are present
try:
    nltk.data.find('corpora/wordnet.zip')
except LookupError:
    nltk.download('wordnet')
    nltk.download('omw-1.4')

# Initialize the Lemmatizer
lemmatizer = WordNetLemmatizer()

# Test: 'better' should become 'good' if pos='a' (adjective)
word = "better"
lemma = lemmatizer.lemmatize(word, pos="a")

print(f"Original: {word}")
print(f"Lemmatized: {lemma}")
```

```python
#Unlike stemming (which just chops off the ends of words), lemmatization uses a dictionary (WordNet) to return the actual base form of a word (the lemma).
#By default, the lemmatizer assumes every word is a Noun.
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Initialize the lemmatizer
lemmatizer = WordNetLemmatizer()

# Simple Noun examples
print(f"cats -> {lemmatizer.lemmatize('cats')}")
print(f"geese -> {lemmatizer.lemmatize('geese')}")
print(f"rocks -> {lemmatizer.lemmatize('rocks')}")
```

```python
# Fails because it assumes 'running' is a noun
print(lemmatizer.lemmatize("running"))  
# Output: 'running' (Incorrect)

# Works because we specify pos="v" (verb)
print(lemmatizer.lemmatize("running", pos="v")) 
# Output: 'run' (Correct)

# Works because we specify pos="a" (adjective)
print(lemmatizer.lemmatize("better", pos="a"))  
# Output: 'good' (Correct)
```
